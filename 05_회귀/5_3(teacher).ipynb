```python
# PoolQC          1453
# MiscFeature     1406
# Alley           1369
# Fence           1179
# FireplaceQu
# df['FireplaceQu'].value_counts()
# Train Set 기준
drop_cols = ['Id', 'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu']
```

```python
# 결측치가 많은 칼럼 드랍
df = df.drop(drop_cols, axis = 1)

# 숫자칼럼의 경우 결측치를 평균으로 대체
num_cols = df.dtypes[df.dtypes != 'object'].index
# num_cols = df.dtypes[df.dtypes != 'str'].index

# 데이터타입 : Object의 경우 여전히 결측치 남아 있음
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
# 결측치가 많은 칼럼 드랍
ID = df_test['Id']

df_test = df_test.drop(drop_cols, axis = 1)

# 숫자칼럼의 경우 결측치를 평균으로 대체
num_cols = df_test.dtypes[df_test.dtypes != 'object'].index
# num_cols = df.dtypes[df.dtypes != 'str'].index

# 데이터타입 : Object의 경우 여전히 결측치 남아 있음
df_test[num_cols] = df_test[num_cols].fillna(df_test[num_cols].mean())
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0].sort_values(ascending=False))
```

```python
print(df_test.isnull().sum()[df_test.isnull().sum() > 0].sort_values(ascending=False))
```

```python
y = df['SalePrice']
X = df.drop(['SalePrice'], axis = 1)
```

```python
X.head()
```

```python
print(y.shape)
print(X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_dum = pd.get_dummies(X)

X_dum.head()
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_dum, y, 
                                    test_size = 0.3, random_state = 42)
```

```python
X_test_dum = pd.get_dummies(df_test)

X_test_dum = X_test_dum.reindex(columns=X_train.columns, fill_value=0)
# print(ID)
# print(X_test_dum)
```

```python
print(X_train.shape)
print(X_test_dum.shape)
```

```python
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.metrics import mean_squared_error

xgb = XGBRegressor(n_estimators = 1000, learning_rate = 0.05, 
                   colsample_bytree = 0.5, subsample=0.8)

lgbm = LGBMRegressor(n_estimators = 1000, learning_rate = 0.05, 
                   colsample_bytree = 0.5, subsample=0.8, n_jobs = -1,
                    verbose = -1)

xgb.fit(X_train, y_train)
xgb_val = xgb.predict(X_val)
xgb_mse = mean_squared_error(y_val, xgb_val)

lgbm.fit(X_train, y_train)
lgbm_val = lgbm.predict(X_val)
lgbm_mse = mean_squared_error(y_val, lgbm_val)

print(xgb_mse)
print(lgbm_mse)
```

```python
pred_xgb = xgb.predict(X_test_dum)
pred_lgbm = lgbm.predict(X_test_dum)
```

```python
res = pd.DataFrame({'ID':ID, 'XGB_SalePrice':pred_xgb, 'LGBM_SalePrice': pred_lgbm})
```
